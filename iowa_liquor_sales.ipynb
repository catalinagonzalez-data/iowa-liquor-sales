## 2. Total sales

## 3. Sales by month

```python
df_2026['month'] = df_2026['ordered_on'].dt.month
sales_2025_jan_oct['month'] = sales_2025_jan_oct['ordered_on'].dt.month

print(df_2026['month'].unique())
print(sales_2025_jan_oct['month'].unique())
```

```python
monthly_2025 = sales_2025_jan_oct.groupby('month')['sales_dollars'].sum()
monthly_2026 = df_2026.groupby('month')['sales_dollars'].sum()

print(monthly_2025)
print(monthly_2026)
```

```python
monthly = pd.DataFrame({'2025': monthly_2025 / 1000000,
                        '2026': monthly_2026 / 1000000})

monthly['change_%'] = (monthly['2026'] - monthly['2025']) / monthly['2025'] * 100

monthly.round(1)
```

```python
plt.figure(figsize=(8, 4))

plt.plot(monthly.index, monthly['2025'], marker='o', color='#B0B7C3', label='2025')
plt.plot(monthly.index, monthly['2026'], marker='o', color='#1F3A5F', label='2026')

plt.title('Iowa liquor sales by month, Jan-Oct: 2025 vs 2026')
plt.xlabel('Month')
plt.ylabel('Sales (USD millions)')
plt.legend()
plt.gca().spines[['top', 'right']].set_visible(False)
plt.xticks(range(1, 11))
plt.show()
```

Both years show the same seasonal trend, with sales building from January through to July. Between January and September, 2026 tracked closely to 2025, just slightly lower in most months. The real difference is October. In 2025 that month hit USD 52.3M, well above the usual USD 31–36M, while October 2026 came in at USD 39.0M, still its best month of the year. In other words, October explains about USD 13.3M of the USD 20.3M gap.

```python
jan_sep = monthly.loc[1:9, ['2025', '2026']].sum()

change_jan_sep = (jan_sep['2026'] - jan_sep['2025']) / jan_sep['2025'] * 100

print(jan_sep.round(1))
print(round(change_jan_sep, 2))
```

From January to September, sales in 2026 fell by only 2.3%: USD 301.5M compared with USD 308.4M in 2025. This is a small decline compared with October, which fell by 25.5%.

With the duplicates removed, the October spike disappears. October is the best month in both years, at about USD 38.5M in 2025 and USD 39.0M in 2026. The two years follow the same pattern, and 2026 is slightly lower in most months. The largest gaps are in January (-8.2%) and April (-5.1%).

## 7. Customer segmentation (RFM)

```python
ref_date = pd.Timestamp('2026-10-30')

rfm = sales_2026_clean.groupby('store_no').agg(
    last_order=('ordered_on', 'max'),
    orders=('invoice_id', 'nunique'),
    sales=('sales_dollars', 'sum')
)

rfm['recency'] = (ref_date - rfm['last_order']).dt.days

print(rfm.shape)
rfm.head()
```

```python
rfm['R'] = pd.cut(rfm['recency'], bins=[-1, 7, 30, 90, 400], labels=[4, 3, 2, 1]).astype(int)
rfm['F'] = pd.qcut(rfm['orders'], 4, labels=[1, 2, 3, 4]).astype(int)
rfm['M'] = pd.qcut(rfm['sales'], 4, labels=[1, 2, 3, 4]).astype(int)

print(rfm['R'].value_counts().sort_index())
rfm.head()
```

```python
def segment(row):
    if row['R'] <= 2 and row['M'] >= 3:
        return 'At risk'
    if row['R'] <= 2:
        return 'Inactive'
    if row['F'] == 4 and row['M'] == 4:
        return 'Top stores'
    return 'Regular'

rfm['segment'] = rfm.apply(segment, axis=1)

summary = rfm.groupby('segment').agg(stores=('sales', 'size'), sales=('sales', 'sum'))
summary['sales'] = summary['sales'] / 1000000
summary['share_%'] = summary['sales'] / summary['sales'].sum() * 100
summary.round(2)
```

```python
summary['stores_%'] = summary['stores'] / summary['stores'].sum() * 100
summary.round(2)
```

```python
store_names = sales_2026_clean.groupby('store_no')['store_name'].last()
rfm['store_name'] = store_names

at_risk = rfm[rfm['segment'] == 'At risk'].sort_values('sales', ascending=False)
at_risk[['store_name', 'last_order', 'orders', 'sales']].head(5)
```

386 stores, 17% of all stores, bring in 67% of 2026 sales, so they should be the distributor’s first priority. 14 stores are at risk: they used to spend well but have not ordered in over a month. The largest, Madella Wine and Spirits in Bettendorf, bought USD 674.8K this year but has not ordered since 13 July. These stores are worth a call to find out why they stopped.
